```python
#Problem 1 — Smart Parking Lot Management System


def parking_system(capacity, logs):
    parked = 0
    peak = 0

    for action in logs:
        if action == "IN":
            parked += 1
        elif action == "OUT":
            parked -= 1

        peak = max(peak, parked)

    status = "Full" if parked >= capacity else "Available"

    print("Currently Parked Vehicles:", parked)
    print("Parking Status:", status)

parking_system(50, ["IN","IN","IN","OUT","IN","IN","OUT"])


#Problem 2 — Online Food Delivery Time Estimator



def delivery_time(distance, traffic, weather):
    time = distance * 5   # base 5 min per km

    if traffic == "High":
        time += 15
    elif traffic == "Medium":
        time += 8

    if weather == "Rainy":
        time += 10

    print("Estimated Delivery Time:", time, "minutes")

delivery_time(8, "High", "Rainy")


#Problem 3 — Movie Theatre Seat Occupancy Analyzer


def seat_occupancy(total_seats, booked):
    booked_count = len(booked)
    occupancy = (booked_count / total_seats) * 100

    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 70:
        status = "Almost Full"
    else:
        status = "Available"

    print("Occupancy:", int(occupancy), "%")
    print("Show Status:", status)

seat_occupancy(200, [1]*150)


#Problem 4 — Cloud Server Load Classification System


def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)

    if avg < 50:
        status = "Normal"
    elif avg <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Average CPU Load:", int(avg), "%")
    print("Server Status:", status)

server_load([45,60,70,85,90])


#Problem 5 — Smart Classroom Resource Usage Monitor


def resource_monitor(resources):
    overused = []

    for k, v in resources.items():
        if v > 8:
            overused.append(k)

    print("Overused Resources:", ",".join(overused))
    print("Energy Alert: Yes" if overused else "Energy Alert: No")

resource_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})


#Problem 6 — Online Event Registration Capacity Controller


def event_registration(capacity, registrations):
    if registrations > capacity:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlist = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)

event_registration(100,105)
```
